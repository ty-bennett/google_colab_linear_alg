```python
import numpy as np
```

```python
A = np.array([[0.,1.],
              [1.,1.],
              [2.,1.]])
b = np.array([1.,1.,2.])
```

```python
x_hat, residuals, rank, singular_vals = np.linalg.lstsq(A,b, rcond=None)

print(x_hat)
```

```python
p = A @ x_hat
r = b - p

print("predicated values = ", p)
print("residual =", r)
print("residual norm (err) =", np.linalg.norm(r))
```

```python
import matplotlib.pyplot as plt
```

```python
t_data = np.array([1.,0.,4.])
y_data = np.array([2.,3.,5.])

A = np.column_stack((t_data, np.ones_like(t_data)))
```

```python
A
```

```python
coeffs, resids, rank, sing_vals = np.linalg.lstsq(A,y_data, rcond=None)
coeffs
```

```python
intercept = coeffs[1]
slope = coeffs[0]
```

```python
t_plot = np.linspace(t_data.min(), t_data.max(), 200)
y_plot = intercept + slope*t_plot
```

```python
plt.scatter(t_data, y_data, color = "blue", label="data")
plt.plot(t_plot, y_plot, color = "red", label="best-fite-line")

plt.axis()
plt.grid()
plt.show()
```
