```python
# Problem 2. Matrix Multiplication
# Let A=(1 2 and B = (2 0
#       -1 3)         4 1)
#
# b. Compute AB and BA and verify answers in Python
```

```python
a = np.array([[1, 2], [-1, 3]])
b = np.array([[2, 0], [4, 1]])
```

```python
print(a@b)
print(b@a)
```

```python
# c. No AB != BA in this example.
# This tells us that the order in which you complete matrix multiplication operations matters since
# AB putting A first, B second does not yield the same result as B first A second.
```
