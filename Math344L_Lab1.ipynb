```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
theta = np.linspace(0, 2*np.pi, 200)
```

```python
X = np.array([
              np.cos(theta),
              np.sin(theta)
              ])
```

```python
X.shape
```

```python
plt.plot(X[0], X[1])
plt.axis("equal")
plt.grid()
plt.show()
```

```python
M = np.array([
              [2,0],
              [0,0]
             ])
```

```python
Y = M@X
```

```python
plt.plot(X[0], X[1], label = "Before")
plt.plot(Y[0], Y[1], label = "After")
#plt.plot(M[0], M[1], M[2])
plt.legend()
plt.axis("equal")
plt.grid()
plt.show()
```
